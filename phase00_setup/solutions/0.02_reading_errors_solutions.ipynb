```python
# ▶ Run this cell first. It defines check() and expect(), which test your answers,
#   and DATA, the path to the course's shared data folder.
import copy, math
from pathlib import Path

DATA = next((p / "data" for p in [Path.cwd(), *Path.cwd().parents] if (p / "data" / "README.md").exists()), Path("data"))

def _same(a, b):
    if hasattr(b, "shape") and hasattr(b, "dtype"):          # numpy arrays
        import numpy as np
        try:
            return np.shape(a) == np.shape(b) and np.allclose(a, b, rtol=1e-5, atol=1e-8)
        except TypeError:
            return False
    if isinstance(b, float) and isinstance(a, (int, float)) and not isinstance(a, bool):
        return math.isclose(a, b, rel_tol=1e-6, abs_tol=1e-9)
    if isinstance(b, (list, tuple)):
        return type(a) is type(b) and len(a) == len(b) and all(_same(x, y) for x, y in zip(a, b))
    if isinstance(b, dict):
        return isinstance(a, dict) and a.keys() == b.keys() and all(_same(a[k], b[k]) for k in b)
    return a == b

def check(fn, cases):
    """Run fn on each (args, expected) pair and report. args is a tuple of arguments."""
    ok = 0
    for args, expected in cases:
        if not isinstance(args, tuple):
            args = (args,)
        shown = ", ".join(repr(a) for a in args)
        if len(shown) > 80:
            shown = shown[:77] + "..."
        try:
            got = fn(*copy.deepcopy(args))
        except Exception as e:
            print(f"  ✗ {fn.__name__}({shown}) raised {type(e).__name__}: {e}")
            continue
        if _same(got, expected):
            ok += 1
        else:
            print(f"  ✗ {fn.__name__}({shown}) returned {got!r}, expected {expected!r}")
    status = "✅ all passed" if ok == len(cases) else "keep going"
    print(f"{fn.__name__}: {ok}/{len(cases)} passed — {status}")

def expect(description, condition):
    """For checks that aren't a simple function call: expect('mean is ~0', abs(m) < 1e-6)."""
    try:
        passed = bool(condition() if callable(condition) else condition)
    except Exception as e:
        print(f"  ✗ {description} — raised {type(e).__name__}: {e}")
        return
    print(("  ✅ " if passed else "  ✗ ") + description + ("" if passed else " — keep going"))
```

## Problem 1

```python
# NameError on the return line: 'lsat' is a typo for 'last'.
def full_name(first, last):
    return first + " " + last

full_name("Ada", "Lovelace")
```

```python
check(full_name, [(("Ada", "Lovelace"), "Ada Lovelace")])
```

## Problem 2

```python
# TypeError: can only concatenate str (not "int") to str. total is a number; convert it (or use an f-string).
def describe_total(total):
    return f"Total: {total} LKR"

describe_total(1500)
```

```python
check(describe_total, [((1500,), "Total: 1500 LKR"), ((0,), "Total: 0 LKR")])
```

## Problem 3

```python
# IndexError: list index out of range. Positions run 0..len-1, so the last one is len(items) - 1 (or items[-1]).
def last_item(items):
    return items[-1]

last_item([3, 8, 5])
```

```python
check(last_item, [(([3, 8, 5],), 5), ((["only"],), "only")])
```

## Problem 4

```python
# KeyError: 'usd'. Dict keys are case-sensitive; normalise the input first.
def get_rate(rates, currency):
    return rates[currency.upper()]

get_rate({"USD": 300.5, "EUR": 325.1}, "usd")
```

```python
check(get_rate, [(({"USD": 300.5, "EUR": 325.1}, "usd"), 300.5), (({"USD": 300.5}, "USD"), 300.5)])
```

## Problem 5 — the sneaky one

The error appears on one line, but the bug is on another. Find the real cause.

```python
# TypeError: 'NoneType' object is not iterable - on the return line, but the bug is the line above:
# list.sort() sorts in place and returns None. Use sorted(), which returns a new list.
def sorted_upper(names):
    ordered = sorted(names)
    return [n.upper() for n in ordered]

sorted_upper(["kasun", "ayesha", "nimal"])
```

```python
check(sorted_upper, [((["kasun", "ayesha", "nimal"],), ["AYESHA", "KASUN", "NIMAL"])])
```

## Problem 6

```python
# ZeroDivisionError: an empty list has length 0. Decide what the answer should be for empty input.
def average(values):
    if not values:
        return 0.0
    return sum(values) / len(values)

average([])
```

```python
check(average, [(([],), 0.0), (([2, 4],), 3.0)])
```

## Problem 7 — parse a number

`int("3.7")` fails. Write `to_whole_number(text)` that turns strings like `"3.7"`, `"12"` and `" 8.2 "` into the whole number, rounded down (3, 12, 8).

```python
# ValueError: invalid literal for int() with base 10: '3.7'. int() on a *string* only accepts whole numbers;
# convert to float first, then to int (which truncates toward zero).
def to_whole_number(text):
    return int(float(text.strip()))
```

```python
check(to_whole_number, [(("3.7",), 3), (("12",), 12), ((" 8.2 ",), 8)])
```
