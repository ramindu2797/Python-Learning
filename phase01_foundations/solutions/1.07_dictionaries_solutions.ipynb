```python
# ▶ Run this cell first. It defines check() and expect(), which test your answers,
#   and DATA, the path to the course's shared data folder.
import copy, math
from pathlib import Path

DATA = next((p / "data" for p in [Path.cwd(), *Path.cwd().parents] if (p / "data" / "README.md").exists()), Path("data"))

def _same(a, b):
    if hasattr(b, "shape") and hasattr(b, "dtype"):          # numpy arrays
        import numpy as np
        try:
            return np.shape(a) == np.shape(b) and np.allclose(a, b, rtol=1e-5, atol=1e-8)
        except TypeError:
            return False
    if isinstance(b, float) and isinstance(a, (int, float)) and not isinstance(a, bool):
        return math.isclose(a, b, rel_tol=1e-6, abs_tol=1e-9)
    if isinstance(b, (list, tuple)):
        return type(a) is type(b) and len(a) == len(b) and all(_same(x, y) for x, y in zip(a, b))
    if isinstance(b, dict):
        return isinstance(a, dict) and a.keys() == b.keys() and all(_same(a[k], b[k]) for k in b)
    return a == b

def check(fn, cases):
    """Run fn on each (args, expected) pair and report. args is a tuple of arguments."""
    ok = 0
    for args, expected in cases:
        if not isinstance(args, tuple):
            args = (args,)
        shown = ", ".join(repr(a) for a in args)
        if len(shown) > 80:
            shown = shown[:77] + "..."
        try:
            got = fn(*copy.deepcopy(args))
        except Exception as e:
            print(f"  ✗ {fn.__name__}({shown}) raised {type(e).__name__}: {e}")
            continue
        if _same(got, expected):
            ok += 1
        else:
            print(f"  ✗ {fn.__name__}({shown}) returned {got!r}, expected {expected!r}")
    status = "✅ all passed" if ok == len(cases) else "keep going"
    print(f"{fn.__name__}: {ok}/{len(cases)} passed — {status}")

def expect(description, condition):
    """For checks that aren't a simple function call: expect('mean is ~0', abs(m) < 1e-6)."""
    try:
        passed = bool(condition() if callable(condition) else condition)
    except Exception as e:
        print(f"  ✗ {description} — raised {type(e).__name__}: {e}")
        return
    print(("  ✅ " if passed else "  ✗ ") + description + ("" if passed else " — keep going"))
```

## Problem 1 — character frequency

`char_frequency(text)` counts every character **except spaces**.

```python
def char_frequency(text):
    counts = {}
    for ch in text:
        if ch != " ":
            counts[ch] = counts.get(ch, 0) + 1
    return counts
```

```python
check(char_frequency, [(("aab b",), {"a": 2, "b": 2}), (("",), {}), (("Aa",), {"A": 1, "a": 1})])
```

## Problem 2 — invert a dict

`invert(d)` swaps keys and values (assume values are unique).

```python
def invert(d):
    out = {}
    for k, v in d.items():
        out[v] = k
    return out
```

```python
check(invert, [(({"a": 1, "b": 2},), {1: "a", 2: "b"}), (({},), {})])
```

## Problem 3 — group by first letter

`group_by_first_letter(words)` → dict from **lower-case** first letter to the list of words, in their original order.

```python
def group_by_first_letter(words):
    groups = {}
    for w in words:
        key = w[0].lower()
        if key not in groups:
            groups[key] = []
        groups[key].append(w)
    return groups
```

```python
check(group_by_first_letter, [
    ((["apple", "Banana", "avocado", "blueberry", "cherry"],), {"a": ["apple", "avocado"], "b": ["Banana", "blueberry"], "c": ["cherry"]}),
    (([],), {}),
])
```

## Problem 4 — total by category

`total_by_category(transactions)` → `{category: total amount}`, totals rounded to 2 decimals.

```python
def total_by_category(transactions):
    totals = {}
    for tx in transactions:
        cat = tx["category"]
        totals[cat] = totals.get(cat, 0) + tx["amount"]
    for cat in totals:
        totals[cat] = round(totals[cat], 2)
    return totals
```

```python
check(total_by_category, [((TXNS,), {"Groceries": 8500.5, "Transport": 2050.0, "Dining": 22700.0}), (([],), {})])
```

## Problem 5 — most common item

`most_common(items)` returns the item that appears most often. If there's a tie, return the one that appeared **first** in the list. Don't use `collections.Counter`.

```python
def most_common(items):
    counts = {}
    for x in items:
        counts[x] = counts.get(x, 0) + 1
    best = None
    for x in counts:                 # dicts keep insertion order = order of first appearance
        if best is None or counts[x] > counts[best]:
            best = x
    return best
```

```python
check(most_common, [((["a", "b", "b", "a", "c", "b"],), "b"), (([3, 1, 1, 3],), 3), ((["x"],), "x")])
```

## Problem 6 — merge counts

`merge_counts(d1, d2)` returns a **new** dict where shared keys have their values added. Don't modify the inputs.

```python
def merge_counts(d1, d2):
    out = dict(d1)
    for k, v in d2.items():
        out[k] = out.get(k, 0) + v
    return out
```

```python
check(merge_counts, [(({"a": 1, "b": 2}, {"b": 3, "c": 4}), {"a": 1, "b": 5, "c": 4}), (({}, {"x": 1}), {"x": 1})])
a, b = {"a": 1}, {"a": 2}
merge_counts(a, b)
expect("inputs unchanged", a == {"a": 1} and b == {"a": 2})
```

## Problem 7 — spend per month

`spend_per_month(transactions)` → `{"YYYY-MM": total}` (rounded to 2 decimals). The month is the first 7 characters of the date.

```python
def spend_per_month(transactions):
    totals = {}
    for tx in transactions:
        month = tx["date"][:7]
        totals[month] = totals.get(month, 0) + tx["amount"]
    return {m: round(t, 2) for m, t in totals.items()}
```

```python
check(spend_per_month, [((TXNS,), {"2026-01": 13550.5, "2026-02": 19700.0})])
```

## Problem 8 ★ — safely read an LLM response

LLM APIs return nested JSON. Write `get_text(response)` that returns `response["choices"][0]["message"]["content"]`, or `None` if **any** level is missing or empty.

```python
def get_text(response):
    choices = response.get("choices")
    if not choices:
        return None
    message = choices[0].get("message")
    if not message:
        return None
    return message.get("content")
```

```python
check(get_text, [
    (({"choices": [{"message": {"content": "Hi!"}}]},), "Hi!"),
    (({"choices": []},), None),
    (({},), None),
    (({"choices": [{"finish_reason": "stop"}]},), None),
    (({"choices": [{"message": {"role": "assistant"}}]},), None),
])
```

## Problem 9 ★ — top merchants

`top_merchants(transactions, n)` returns a list of `(merchant, total)` tuples for the `n` merchants with the highest total spend, highest first. Totals rounded to 2 decimals.

```python
def top_merchants(transactions, n):
    totals = {}
    for tx in transactions:
        totals[tx["merchant"]] = totals.get(tx["merchant"], 0) + tx["amount"]
    ranked = sorted(totals.items(), key=lambda kv: kv[1], reverse=True)
    return [(m, round(t, 2)) for m, t in ranked[:n]]
```

```python
check(top_merchants, [((TXNS, 2), [("Ministry of Crab", 18500.0), ("Keells", 8500.5)]), ((TXNS, 1), [("Ministry of Crab", 18500.0)])])
```

## Problem 10 ★ — anagram groups

`anagram_groups(words)` groups words that are anagrams of each other. Return a list of groups (lists), in order of each group's first appearance. Hint: what key is identical for all anagrams?

```python
def anagram_groups(words):
    groups = {}
    for w in words:
        key = "".join(sorted(w))
        groups.setdefault(key, []).append(w)
    return list(groups.values())
```

```python
check(anagram_groups, [
    ((["listen", "silent", "enlist", "google", "gogole", "cat"],), [["listen", "silent", "enlist"], ["google", "gogole"], ["cat"]]),
    (([],), []),
])
```
