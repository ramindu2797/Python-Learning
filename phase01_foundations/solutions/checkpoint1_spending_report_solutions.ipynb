```python
# ▶ Run this cell first. It defines check() and expect(), which test your answers,
#   and DATA, the path to the course's shared data folder.
import copy, math
from pathlib import Path

DATA = next((p / "data" for p in [Path.cwd(), *Path.cwd().parents] if (p / "data" / "README.md").exists()), Path("data"))

def _same(a, b):
    if hasattr(b, "shape") and hasattr(b, "dtype"):          # numpy arrays
        import numpy as np
        try:
            return np.shape(a) == np.shape(b) and np.allclose(a, b, rtol=1e-5, atol=1e-8)
        except TypeError:
            return False
    if isinstance(b, float) and isinstance(a, (int, float)) and not isinstance(a, bool):
        return math.isclose(a, b, rel_tol=1e-6, abs_tol=1e-9)
    if isinstance(b, (list, tuple)):
        return type(a) is type(b) and len(a) == len(b) and all(_same(x, y) for x, y in zip(a, b))
    if isinstance(b, dict):
        return isinstance(a, dict) and a.keys() == b.keys() and all(_same(a[k], b[k]) for k in b)
    return a == b

def check(fn, cases):
    """Run fn on each (args, expected) pair and report. args is a tuple of arguments."""
    ok = 0
    for args, expected in cases:
        if not isinstance(args, tuple):
            args = (args,)
        shown = ", ".join(repr(a) for a in args)
        if len(shown) > 80:
            shown = shown[:77] + "..."
        try:
            got = fn(*copy.deepcopy(args))
        except Exception as e:
            print(f"  ✗ {fn.__name__}({shown}) raised {type(e).__name__}: {e}")
            continue
        if _same(got, expected):
            ok += 1
        else:
            print(f"  ✗ {fn.__name__}({shown}) returned {got!r}, expected {expected!r}")
    status = "✅ all passed" if ok == len(cases) else "keep going"
    print(f"{fn.__name__}: {ok}/{len(cases)} passed — {status}")

def expect(description, condition):
    """For checks that aren't a simple function call: expect('mean is ~0', abs(m) < 1e-6)."""
    try:
        passed = bool(condition() if callable(condition) else condition)
    except Exception as e:
        print(f"  ✗ {description} — raised {type(e).__name__}: {e}")
        return
    print(("  ✅ " if passed else "  ✗ ") + description + ("" if passed else " — keep going"))
```

## Task 1 — total per category

`total_by_category(txns)` → `{category: total}`, totals rounded to 2 decimals.

```python
def total_by_category(txns):
    # plan: walk the transactions, add each amount to its category's running total, round at the end
    totals = {}
    for t in txns:
        totals[t["category"]] = totals.get(t["category"], 0) + t["amount"]
    return {c: round(v, 2) for c, v in totals.items()}
```

```python
check(total_by_category, [((SAMPLE,), {"Groceries": 10500.0, "Transport": 2500.0, "Shopping": 12000.0})])
```

## Task 2 — top merchants

`top_merchants(txns, n=3)` → list of `(merchant, total)` tuples, biggest total first, totals rounded to 2 decimals.

```python
def top_merchants(txns, n=3):
    # plan: total per merchant (same as task 1), sort the (merchant, total) pairs by total descending, keep n
    totals = {}
    for t in txns:
        totals[t["merchant"]] = totals.get(t["merchant"], 0) + t["amount"]
    ranked = sorted(totals.items(), key=lambda kv: kv[1], reverse=True)
    return [(m, round(v, 2)) for m, v in ranked[:n]]
```

```python
check(top_merchants, [((SAMPLE,), [("Odel", 12000.0), ("Keells", 10500.0), ("PickMe", 2500.0)]), ((SAMPLE, 1), [("Odel", 12000.0)])])
```

## Task 3 — months over budget

`months_over_budget(txns, budget)` → sorted list of `"YYYY-MM"` strings whose total spend is **strictly greater** than `budget`.

```python
def months_over_budget(txns, budget):
    # plan: total per month (first 7 chars of date), keep months above budget, sort
    per_month = {}
    for t in txns:
        m = t["date"][:7]
        per_month[m] = per_month.get(m, 0) + t["amount"]
    return sorted(m for m, total in per_month.items() if total > budget)
```

```python
check(months_over_budget, [((SAMPLE, 8000), ["2026-01", "2026-02"]), ((SAMPLE, 13500), []), ((SAMPLE, 0), ["2026-01", "2026-02", "2026-03"])])
```

## Task 4 — formatted report

`format_report(txns)` returns one string (lines joined with `"\n"`):

- a header line: `f"{'Category':<15}{'Total (LKR)':>15}{'Share':>8}"`
- one line per category, **biggest total first**: `f"{category:<15}{total:>15,.2f}{share:>8.1%}"` where `share` is the category total divided by the grand total
- a final line: `f"{'TOTAL':<15}{grand_total:>15,.2f}{1:>8.1%}"`

Reuse `total_by_category`.

```python
def format_report(txns):
    # plan: category totals -> grand total -> sort categories by total desc -> build lines -> join
    totals = total_by_category(txns)
    grand = sum(totals.values())
    lines = [f"{'Category':<15}{'Total (LKR)':>15}{'Share':>8}"]
    for cat, total in sorted(totals.items(), key=lambda kv: kv[1], reverse=True):
        lines.append(f"{cat:<15}{total:>15,.2f}{total / grand:>8.1%}")
    lines.append(f"{'TOTAL':<15}{grand:>15,.2f}{1:>8.1%}")
    return "\n".join(lines)
```

```python
expected = (
    "Category           Total (LKR)   Share\n"
    "Shopping             12,000.00   48.0%\n"
    "Groceries            10,500.00   42.0%\n"
    "Transport             2,500.00   10.0%\n"
    "TOTAL                25,000.00  100.0%"
)
check(format_report, [((SAMPLE,), expected)])
```
