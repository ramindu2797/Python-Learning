```python
# ▶ Run this cell first. It defines check() and expect(), which test your answers,
#   and DATA, the path to the course's shared data folder.
import copy, math
from pathlib import Path

DATA = next((p / "data" for p in [Path.cwd(), *Path.cwd().parents] if (p / "data" / "README.md").exists()), Path("data"))

def _same(a, b):
    if hasattr(b, "shape") and hasattr(b, "dtype"):          # numpy arrays
        import numpy as np
        try:
            return np.shape(a) == np.shape(b) and np.allclose(a, b, rtol=1e-5, atol=1e-8)
        except TypeError:
            return False
    if isinstance(b, float) and isinstance(a, (int, float)) and not isinstance(a, bool):
        return math.isclose(a, b, rel_tol=1e-6, abs_tol=1e-9)
    if isinstance(b, (list, tuple)):
        return type(a) is type(b) and len(a) == len(b) and all(_same(x, y) for x, y in zip(a, b))
    if isinstance(b, dict):
        return isinstance(a, dict) and a.keys() == b.keys() and all(_same(a[k], b[k]) for k in b)
    return a == b

def check(fn, cases):
    """Run fn on each (args, expected) pair and report. args is a tuple of arguments."""
    ok = 0
    for args, expected in cases:
        if not isinstance(args, tuple):
            args = (args,)
        shown = ", ".join(repr(a) for a in args)
        if len(shown) > 80:
            shown = shown[:77] + "..."
        try:
            got = fn(*copy.deepcopy(args))
        except Exception as e:
            print(f"  ✗ {fn.__name__}({shown}) raised {type(e).__name__}: {e}")
            continue
        if _same(got, expected):
            ok += 1
        else:
            print(f"  ✗ {fn.__name__}({shown}) returned {got!r}, expected {expected!r}")
    status = "✅ all passed" if ok == len(cases) else "keep going"
    print(f"{fn.__name__}: {ok}/{len(cases)} passed — {status}")

def expect(description, condition):
    """For checks that aren't a simple function call: expect('mean is ~0', abs(m) < 1e-6)."""
    try:
        passed = bool(condition() if callable(condition) else condition)
    except Exception as e:
        print(f"  ✗ {description} — raised {type(e).__name__}: {e}")
        return
    print(("  ✅ " if passed else "  ✗ ") + description + ("" if passed else " — keep going"))
```

## Problem 1 — min and max together

`min_max(values)` returns a **tuple** `(smallest, largest)` in a single pass (no `min()`/`max()`). Then, in a scratch cell, call it and unpack the result into two variables.

```python
def min_max(values):
    lo = hi = values[0]
    for v in values[1:]:
        if v < lo:
            lo = v
        if v > hi:
            hi = v
    return lo, hi

lo, hi = min_max([4, 9, 1])
```

```python
check(min_max, [(([4, 9, 1],), (1, 9)), (([5],), (5, 5)), (([-2, -8, 0],), (-8, 0))])
```

## Problem 2 — any duplicates?

`has_duplicates(items)` in one line.

```python
def has_duplicates(items):
    return len(set(items)) != len(items)
```

```python
check(has_duplicates, [(([1, 2, 3],), False), (([1, 2, 1],), True), (([],), False)])
```

## Problem 3 — unique emails

`count_unique_emails(emails)` counts distinct addresses, ignoring case and surrounding spaces.

```python
def count_unique_emails(emails):
    return len({e.strip().lower() for e in emails})
```

```python
check(count_unique_emails, [((["a@x.com", "A@x.com ", "b@x.com"],), 2), (([],), 0)])
```

## Problem 4 — joiners, leavers, stayers

`staff_changes(this_month, last_month)` takes two lists of employee IDs and returns a tuple of three **sorted lists**: `(joiners, leavers, stayers)`.

```python
def staff_changes(this_month, last_month):
    now, before = set(this_month), set(last_month)
    return sorted(now - before), sorted(before - now), sorted(now & before)
```

```python
check(staff_changes, [
    ((["E01", "E02", "E03", "E05"], ["E01", "E02", "E04"]), (["E03", "E05"], ["E04"], ["E01", "E02"])),
    (([], ["E1"]), ([], ["E1"], [])),
])
```

## Problem 5 — pair that sums to a target

`has_pair_with_sum(nums, target)` returns `True` if any **two different positions** hold numbers adding to `target`. Walk the list once, remembering what you've seen in a set. (You'll meet this pattern again in Phase 2.)

```python
def has_pair_with_sum(nums, target):
    seen = set()
    for x in nums:
        if target - x in seen:
            return True
        seen.add(x)
    return False
```

```python
check(has_pair_with_sum, [(([2, 7, 11, 15], 9), True), (([3, 3], 6), True), (([3], 6), False), (([1, 2, 3], 7), False)])
```

## Problem 6 ★ — common to all lists

`common_to_all(*lists)` returns a sorted list of items present in **every** list given. (`*lists` collects any number of arguments into a tuple — covered properly in 1.09.) No lists → `[]`.

```python
def common_to_all(*lists):
    if not lists:
        return []
    common = set(lists[0])
    for lst in lists[1:]:
        common &= set(lst)
    return sorted(common)
```

```python
check(common_to_all, [(([1, 2, 3], [2, 3, 4], [3, 2, 9]), [2, 3]), (([1], [2]), []), ((), [])])
```

## Problem 7 ★ — spend grid with tuple keys

`spend_grid(transactions)` returns a dict whose keys are `(month, category)` tuples and whose values are the totals. Month = first 7 characters of the date.

```python
def spend_grid(transactions):
    grid = {}
    for tx in transactions:
        key = (tx["date"][:7], tx["category"])
        grid[key] = grid.get(key, 0) + tx["amount"]
    return grid
```

```python
sample = [
    {"date": "2026-01-03", "category": "Dining", "amount": 100.0},
    {"date": "2026-01-20", "category": "Dining", "amount": 50.0},
    {"date": "2026-02-01", "category": "Dining", "amount": 70.0},
]
check(spend_grid, [((sample,), {("2026-01", "Dining"): 150.0, ("2026-02", "Dining"): 70.0})])
```
